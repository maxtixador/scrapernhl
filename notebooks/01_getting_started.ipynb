## Method 1: Functional API (Recommended for Beginners)

The simplest way to use ScraperNHL is with the `scrape()` function:

````python
from scrapernhl import scrape
import pandas as pd

pd.set_option('display.max_columns', None)

# Get player stats from the QMJHL
stats = scrape('qmjhl', 'stats', season=90, position='skaters', limit=10)

print(f"Got {len(stats)} players")
print("\nTop 10 players:")
display(stats[['name', 'team_code', 'goals', 'assists', 'points']].head(10))
````

````python
stats
````

### Available Data Types

You can request different types of data:

1. **'pbp'** - Play-by-play data for a specific game
2. **'stats'** - Player statistics for skaters or goalies
3. **'schedule'** - Schedule for a team or entire league
4. **'roster'** - Team roster with player details
5. **'standings'** - Current league standings

````python
# Get schedule for a team
schedule = scrape('ohl', 'schedule', season=83, team_id=16)

print(f"Games in schedule: {len(schedule)}")
display(schedule[['date', 'homeTeam', 'awayTeam']].head())
````

## Method 2: Object-Oriented API (Advanced Usage)

For more control, you can create a `HockeyScraper` object:

### ✨ NEW: Automatic Bootstrap Fetching

When you create a scraper for non-NHL leagues, bootstrap data (teams, seasons, divisions) is **automatically fetched**!

````python
from scrapernhl import HockeyScraper


for league in ['ohl', 'whl', 'qmjhl', 'ahl', 'pwhl']:
    # Create a scraper for the OHL
    scraper = HockeyScraper(league)

    # Bootstrap data is automatically fetched!
    print(f"Bootstrap auto-fetched on initialization:")
    print(f"  League ID: {scraper.current_league_id}")
    print(f"  Season ID: {scraper.current_season_id}")
    print(f"  Teams: {len(scraper.teams)}")

    # Access team data easily
    print(f"\nFirst 5 teams:")
    for team in scraper.teams[:5]:
        print(f"  - {team['name']} (ID: {team['id']})")

    # Use methods to get data
    stats = scraper.player_stats(season=83, position='skaters', limit=15)
    schedule = scraper.schedule(season=83, team='all')  # 'all' or team_id
    standings = scraper.standings(season=83)

    print(f"\nData retrieved:")
    print(f"  Stats: {len(stats)} rows")
    print(f"  Schedule: {len(schedule)} games")
    print(f"  Standings: {len(standings)} teams")

    bootstrap_keys = scraper.bootstrap_data.keys()
    print(f"\nBootstrap data keys: {bootstrap_keys}")
````

````python
bs_data = {}
for league in ['ohl', 'whl', 'qmjhl', 'ahl', 'pwhl']:
    scraper = HockeyScraper(league)
    bs_data[league] = scraper.bootstrap_data
    
bs_data
````

````python
import json

# save to a file
bs_data_json = json.dumps(bs_data, indent=4)
with open('bootstrap_data.json', 'w') as f:
    f.write(bs_data_json)
````

````python
pd.json_normalize(scraper.bootstrap_data["seasons"])
````

````python
# Manually fetch bootstrap with specific parameters
ahl = HockeyScraper('ahl')

# Get bootstrap for a specific season
bootstrap = ahl.bootstrap(season='90', page_name='scorebar')

print(f"Manual bootstrap fetch:")
print(f"  Season ID: {bootstrap['current_season_id']}")
print(f"  Teams: {len(bootstrap['teams'])}")

# Note: NHL doesn't support bootstrap
try:
    nhl = HockeyScraper('nhl')
    nhl_bootstrap = nhl.bootstrap()
except NotImplementedError as e:
    print(f"\nNHL doesn't support bootstrap: {e}")
````

### Manual Bootstrap Method

You can also manually fetch bootstrap with custom parameters:

````python
# Bootstrap properties provide quick access to common data
pwhl = HockeyScraper('pwhl')

print("Convenience properties:")
print(f"  League ID: {pwhl.current_league_id}")
print(f"  Season ID: {pwhl.current_season_id}")
print(f"  Number of teams: {len(pwhl.teams)}")

# Full bootstrap data is also available
print(f"\nFull bootstrap has {len(pwhl.bootstrap_data)} keys:")
print(f"  Available keys: {list(pwhl.bootstrap_data.keys())[:10]}...")

# Access specific data from bootstrap
divisions = pwhl.bootstrap_data.get('divisions', [])
seasons = pwhl.bootstrap_data.get('regularSeasons', [])
print(f"\n  Divisions: {len(divisions)}")
print(f"  Regular seasons: {len(seasons)}")
````

## Understanding the Output

All methods return **pandas DataFrames** - this makes it easy to:
- Filter and sort data
- Export to CSV, Excel, or JSON
- Perform analysis
- Create visualizations

````python
# Example: Filter and sort
stats = scrape('qmjhl', 'stats', season=90, position='skaters', limit=50)

# Get players with more than 20 goals
goal_scorers = stats[stats['G'] > 20].sort_values('G', ascending=False)

print(f"Found {len(goal_scorers)} players with 20+ goals")
display(goal_scorers[['playerName', 'teamCode', 'G', 'A', 'PTS']].head(10))
````

## League-Specific Parameters

Each league and data type accepts specific parameters. Here are the most common ones:

### Player Stats
- `season`: Season ID (e.g., 90 for QMJHL 2024-25)
- `position`: 'skaters' or 'goalies'
- `team`: Team ID or 'all'
- `limit`: Number of results to return

### Schedule
- `season`: Season ID
- `team_id`: Team ID (-1 for all teams)
- `month`: Month filter (-1 for all months)

### Play-by-Play
- `game_id`: The game ID to scrape
- `nhlify`: Apply NHL-style transformations (default: True)

### Roster
- `team_id`: Team ID
- `season`: Season ID

### Standings
- `season`: Season ID
